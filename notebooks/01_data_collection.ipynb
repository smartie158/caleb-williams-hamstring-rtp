```python
##how many consecutive hamstring-report weeks
```

```python
hamstring_all["previous_week"] = (
    hamstring_all
    .groupby("player_season")["week"]
    .shift(1)
)

hamstring_all["week_gap"] = (
    hamstring_all["week"] - hamstring_all["previous_week"]
)
```

```python
##calculate difference between consecutive hamstring report weeks. if it is listed as week 3, week 4, week 5; probably a continuous injury episode. but if it is week 3, week 4... week 12, week 13; probably not the same injury episode. This will help determine how many unique hamstring injuries are represented in the dataset.
```

```python
##define new episode whenever the player's hamstring reports are separated by more than one NFL week.
```

```python
hamstring_all["new_episode"] = (
    hamstring_all["week_gap"].isna()
    | (hamstring_all["week_gap"] > 1)
).astype(int)
```

```python
hamstring_all["episode_number"] = (
    hamstring_all
    .groupby("player_season")["new_episode"]
    .cumsum()
)
```

```python
hamstring_all["injury_id"] = (
    hamstring_all["player_season"]
    + "_"
    + hamstring_all["episode_number"].astype(str)
)
```

```python
## the '>1 week = new injury' rule is subjective. A player could disappear from the injury report for a week or more and then aggrevate the same hamstring. or reporting practices could create gaps. Need to document the limitation rather than pretending we know the medicial diagnosis from public data. 
```

```python
# Alternative episode definition:
# A new injury episode begins after a gap of more than 2 weeks

hamstring_all["new_episode_2wk"] = (
    hamstring_all["week_gap"].isna()
    | (hamstring_all["week_gap"] > 2)
).astype(int)

hamstring_all["episode_number_2wk"] = (
    hamstring_all
    .groupby("player_season")["new_episode_2wk"]
    .cumsum()
)

hamstring_all["injury_id_2wk"] = (
    hamstring_all["player_season"]
    + "_"
    + hamstring_all["episode_number_2wk"].astype(str)
)

hamstring_all["injury_id_2wk"].nunique()
```

```python
## Adding NFL Schedule Data
## Player statistics tells when a QB recrded game statistics but the absence, of a statistical row does not indicate that a game was missed because the team might have had a bye week. NFL schedule data is used to identify the actual regular-season games available to each QB during an injury game
```

```python
schedule_url = (
    "https://github.com/nflverse/nfldata/raw/master/data/games.csv"
)

schedule = pd.read_csv(schedule_url)

print(schedule.shape)
schedule.columns.tolist()
```

```python
schedule[
    [
        "season",
        "week",
        "game_type",
        "home_team",
        "away_team"
    ]
].head()
```

```python
def get_qb_team(player_id, season, start_week):
    rows = player_stats.loc[
        (player_stats["player_id"].astype(str) == str(player_id))
        & (player_stats["season"] == season)
        & (player_stats["week"] <= start_week)
    ].sort_values("week")

    if len(rows) == 0:
        return None

    return rows.iloc[-1]["recent_team"]


qb_episode_summary["team"] = qb_episode_summary.apply(
    lambda row: get_qb_team(
        row["gsis_id"],
        row["season"],
        row["start_week"]
    ),
    axis=1
)

qb_episode_summary[
    ["player", "season", "start_week", "team"]
].sort_values(["season", "start_week"])
```

```python
##determine each team's game weeks
```

```python
regular_schedule = schedule[
    schedule["game_type"] == "REG"
].copy()
```

```python
def get_team_game_weeks(team, season):
    if pd.isna(team):
        return []

    games = regular_schedule.loc[
        (regular_schedule["season"] == season)
        & (
            (regular_schedule["home_team"] == team)
            | (regular_schedule["away_team"] == team)
        ),
        "week"
    ]

    return sorted(
        games.dropna().astype(int).unique().tolist()
    )


qb_episode_summary["team_game_weeks"] = qb_episode_summary.apply(
    lambda row: get_team_game_weeks(
        row["team"],
        row["season"]
    ),
    axis=1
)
```

```python
##Constructing the Return-to-Play Timeline
## Injury report duration does not represent time lost from the season. To create a more accurate outcome, scheduled team games are compared with actual QB participation beginning witht he first week of each hamstring injury. This can distinguish injury report form actual games missed
```

```python
##First game QB played after injury report began
```

```python
def injury_game_timeline(row):
    team_weeks = row["team_game_weeks"]
    stat_weeks = row["stat_weeks"]

    if not isinstance(team_weeks, list) or len(team_weeks) == 0:
        return []

    return [
        week for week in team_weeks
        if week >= row["start_week"]
    ]


qb_episode_summary["injury_game_weeks"] = qb_episode_summary.apply(
    injury_game_timeline,
    axis=1
)

qb_episode_summary[
    [
        "player",
        "season",
        "start_week",
        "team",
        "injury_game_weeks",
        "stat_weeks"
    ]
].sort_values(["season", "start_week"])
```

```python
def get_first_game_back(row):
    playable_weeks = set(row["stat_weeks"])
    
    return next(
        (
            week
            for week in row["injury_game_weeks"]
            if week in playable_weeks
        ),
        pd.NA
    )


qb_episode_summary["first_game_back"] = (
    qb_episode_summary.apply(get_first_game_back, axis=1)
    .astype("Int64")
)

qb_episode_summary[
    [
        "player",
        "season",
        "start_week",
        "initial_report_status",
        "injury_game_weeks",
        "stat_weeks",
        "first_game_back"
    ]
].sort_values(["season", "start_week"])
```

```python
def calculate_games_missed(row):
    injury_games = row["injury_game_weeks"]
    first_back = row["first_game_back"]

    if not isinstance(injury_games, list) or len(injury_games) == 0:
        return pd.NA

    if pd.isna(first_back):
        return pd.NA

    games_before_return = [
        week
        for week in injury_games
        if week < first_back
    ]

    return len(games_before_return)


qb_episode_summary["games_missed"] = (
    qb_episode_summary.apply(calculate_games_missed, axis=1)
    .astype("Int64")
)
```

```python
qb_episode_summary["weeks_to_return"] = (
    qb_episode_summary["first_game_back"]
    - qb_episode_summary["start_week"]
).astype("Int64")
```

```python
##replacce/calculate the old first_game_back
```

```python
def get_true_first_game_back(row):
    missed_weeks = row["missed_game_weeks"]
    
    # No games missed = no RTP event
    if not isinstance(missed_weeks, list) or len(missed_weeks) == 0:
        return pd.NA
    
    last_missed_week = max(missed_weeks)
    stat_weeks = sorted(row["stat_weeks"])
    
    return next(
        (
            week
            for week in stat_weeks
            if week > last_missed_week
        ),
        pd.NA
    )


qb_episode_summary["first_game_back"] = (
    qb_episode_summary.apply(
        get_true_first_game_back,
        axis=1
    )
    .astype("Int64")
)
```

```python
##recalculate the weeks_to_return, because the older first_game_back calculation measured from start_week
```

```python
###seperate calendar style variable for validation
qb_episode_summary["weeks_to_return"] = (
    qb_episode_summary["first_game_back"]
    - qb_episode_summary["start_week"]
).astype("Int64")
```

```python
# Step 42 — classify no-return episodes

def classify_no_return(row):
    if row["games_missed"] == 0:
        return "no_games_missed"

    if pd.notna(row["first_game_back"]):
        return "observed_return"

    missed = row["missed_game_weeks"]
    team_weeks = row["team_game_weeks"]

    if not isinstance(missed, list) or len(missed) == 0:
        return "unknown"

    if not isinstance(team_weeks, list) or len(team_weeks) == 0:
        return "unknown"

    last_missed = max(missed)
    later_team_games = [w for w in team_weeks if w > last_missed]

    if len(later_team_games) == 0:
        return "season_ended_no_return_observable"

    return "later_games_exist_no_return"


qb_episode_summary["rtp_status"] = qb_episode_summary.apply(
    classify_no_return,
    axis=1
)

qb_episode_summary[
    [
        "player",
        "season",
        "start_week",
        "missed_game_weeks",
        "games_missed",
        "first_game_back",
        "rtp_status"
    ]
].sort_values(
    ["rtp_status", "season", "start_week"]
)
```

```python
# Step 51 — create final RTP outcome

def get_final_rtp_outcome(row):

    # Manual validation takes priority
    if pd.notna(row["validated_outcome"]):
        return row["validated_outcome"]

    # Player never missed a game
    if row["rtp_status"] == "no_games_missed":
        return "no_games_missed"

    # Player missed time and we observed a later game appearance
    if row["rtp_status"] == "observed_return":
        return "observed_return"

    # Season ended before return could be observed
    if row["rtp_status"] == "season_ended_no_return_observable":
        return "right_censored"

    return "unresolved"


qb_episode_final["final_rtp_outcome"] = (
    qb_episode_final.apply(get_final_rtp_outcome, axis=1)
)

qb_episode_final["final_rtp_outcome"].value_counts(dropna=False)
```

```python
# Step 73 — create severity comparison groups

def assign_severity_group(row):
    grade = row["reported_grade"]
    severity = row["reported_severity"]

    # Treat pandas.NA, NaN, and other non-string values as missing
    grade = grade if isinstance(grade, str) else None
    severity = severity if isinstance(severity, str) else None

    if grade == "Grade 1":
        return "grade_1"

    if severity == "mild":
        return "mild_grade_unknown"

    if grade == "Grade 2" or severity == "moderate":
        return "moderate"

    if severity in {"severe", "severe/surgical"}:
        return "severe"

    if severity == "moderate/severe":
        return "moderate_severe_unknown_grade"

    return "unknown"


qb_episode_final["severity_group"] = (
    qb_episode_final.apply(assign_severity_group, axis=1)
)


print("Severity groups:")
print(
    qb_episode_final["severity_group"]
    .value_counts(dropna=False)
)

print("\nCases with severity evidence:")
print(
    qb_episode_final.loc[
        qb_episode_final["severity_group"] != "unknown",
        [
            "player",
            "season",
            "reported_grade",
            "reported_severity",
            "severity_group",
            "games_missed",
            "final_rtp_outcome"
        ]
    ]
    .sort_values(["severity_group", "season"])
    .to_string(index=False)
)
```

```python
# Step 88 — rebuild final modeling datasets

# Stage 1:
# Predict whether the injury episode is associated with
# missing at least one scheduled game.
stage1_data = qb_episode_final.loc[
    qb_episode_final["eligible_missed_game_model"]
].copy()


# Stage 2:
# Describe RTP duration only for clean observed-return cases.
stage2_data = qb_episode_final.loc[
    qb_episode_final["eligible_rtp_duration_model"]
].copy()

stage2_data["rtp_games_missed"] = (
    stage2_data["games_missed"].astype("Int64")
)


print("Master episodes:", len(qb_episode_final))
print("Stage 1 observations:", len(stage1_data))
print("Stage 2 observations:", len(stage2_data))

print("\nStage 1 target:")
print(
    stage1_data["missed_game_target"]
    .value_counts()
    .sort_index()
)

print("\nStage 2 RTP games missed:")
print(
    stage2_data["rtp_games_missed"]
    .value_counts()
    .sort_index()
)
```

```python
# Step 89 — final data integrity checks

# Master dataset
assert len(qb_episode_final) == 29, "Master dataset should contain 29 episodes"

assert (
    qb_episode_final
    .duplicated(["player", "season", "start_week"])
    .sum()
    == 0
), "Duplicate injury episodes found"


# Stage 1
assert len(stage1_data) == 24, "Stage 1 should contain 24 observations"

assert (
    stage1_data["missed_game_target"]
    .isna()
    .sum()
    == 0
), "Stage 1 target contains missing values"

assert set(
    stage1_data["missed_game_target"].unique()
) == {0, 1}, "Unexpected Stage 1 target values"


# Stage 2
assert len(stage2_data) == 8, "Stage 2 should contain 8 observations"

assert (
    stage2_data["rtp_games_missed"]
    .isna()
    .sum()
    == 0
), "Stage 2 target contains missing values"


# Severity research
assert len(severity_validation) == 29

assert (
    severity_validation["severity_evidence"]
    .isna()
    .sum()
    == 0
)


print("✓ Master dataset passed")
print("✓ Stage 1 dataset passed")
print("✓ Stage 2 dataset passed")
print("✓ Severity validation passed")
print("\nAll final data integrity checks passed.")
```
